```python
import pandas as pd
import math
```

```python
data = {
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class': ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']
}

df = pd.DataFrame(data)

print("Existing Data:")
print(df)
```

```python
new_brightness = float(input("Enter Brightness of new entry: "))
new_saturation = float(input("Enter Saturation of new entry: "))

print("\nNew Entry:")
print("Brightness:", new_brightness)
print("Saturation:", new_saturation)
```

```python
distances = []

for i in range(len(df)):
    brightness = df.loc[i, 'Brightness']
    saturation = df.loc[i, 'Saturation']

    distance = math.sqrt(
        (brightness - new_brightness) ** 2 +
        (saturation - new_saturation) ** 2
    )

    distances.append(distance)

df['Distance'] = distances

print("Distances from New Entry:")
print(df)
```

```python
print("Distance from New Entry to Each Existing Point:\n")

for i in range(len(df)):
    print(
        "Point", i + 1,
        "(", df.loc[i, 'Brightness'],
        ",", df.loc[i, 'Saturation'], ")",
        "Class:", df.loc[i, 'Class'],
        "Distance:", round(df.loc[i, 'Distance'], 2)
    )
```

```python
sorted_df = df.sort_values(by='Distance')

print("Points arranged from nearest to farthest:")
print(sorted_df)
```

```python
nearest = sorted_df.iloc[0]

print("Nearest Point:")
print("Brightness:", nearest['Brightness'])
print("Saturation:", nearest['Saturation'])
print("Class:", nearest['Class'])
print("Distance:", round(nearest['Distance'], 2))
```
